# Total Energy Consumption
Get the total energy Consumption:  
$\forall bus \sum_{bus} energy\space consumed$ 

```python
import duckdb
import pandas as pd
```

```python
DATABASE_PATH = "../../database/eBuS.duckdb"
VEHICLE_DICT = {"Ebusco2.2electric12m": "Ebusco 2.2", "SolarisUrbino12electric": "Urbino 12", "SolarisUrbino18electric": "Urbino 18"}
```

```python
con = duckdb.connect(DATABASE_PATH)

df = con.execute("""
    SELECT 
        tripinfo_id,
        battery_totalEnergyConsumed,
        battery_totalEnergyRegenerated, 
        tripinfo_routeLength,
        tripinfo_vType,
        scenario,
        seed
    FROM multirun_tripinfo
    WHERE scenario IN ('summer', 'winter', 'reduced', 'increased')
""").fetchdf()

con.close()

for scenario in [
    'summer', 'winter', 'reduced', 'increased'
    ]:

    scenario_consumed = []

    for seed in [65, 67, 68, 70, 75]:
        subset = df[
            (df["scenario"] == scenario) &
            (df["seed"] == seed)
        ].copy()

        energy_expended = (subset['battery_totalEnergyConsumed'].sum() - subset['battery_totalEnergyRegenerated'].sum()) / 1000000
        scenario_consumed.append(energy_expended)
        print(f"Sum Energy Consumed {scenario.upper()}, {seed}: {energy_expended} MWh")
    ec_df = pd.DataFrame(scenario_consumed)
    print(f"{scenario}: {ec_df.describe()}")
```
