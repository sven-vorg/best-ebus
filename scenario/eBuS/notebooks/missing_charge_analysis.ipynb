# Analysis Notebook for missing charging Events

```python
# AI-generated on 2026-09-10

import xml.etree.ElementTree as ET

tree = ET.parse(r"G:\Dokumente\Studium\FU Berlin\BeSTeBuS\best-ebus\scenario\sumo\electric\e_routes.rou.xml")
root = tree.getroot()

route_dict = {}

for route in root.findall("route"):

    firstStops = []
    lastStops = []

    stops = route.findall("stop")

    previousStop = stops[0]
    tripID = previousStop.get("tripId")

    for busStop in stops:

        currentTripID = busStop.get("tripId")

        if currentTripID != tripID:
            firstStops.append(busStop)
            lastStops.append(previousStop)

            tripID = currentTripID

        previousStop = busStop

    lastStops.append(previousStop)

    # Save the stops for this route
    route_id = route.get("id")

    route_dict[route_id] = {
        "firstStops": firstStops,
        "lastStops": lastStops
    }

for route_id, stops in route_dict.items():

    print("Route:", route_id)

    print("First stops:")
    for stop in stops["firstStops"]:
        print(stop.get("busStop"))

    print("Last stops:")
    for stop in stops["lastStops"]:
        print(stop.get("busStop"))
```
